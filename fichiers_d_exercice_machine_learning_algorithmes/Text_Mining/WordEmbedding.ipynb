```python
from gensim.models import Word2Vec
```

```python
# Le corps d'entrainement du modèle Word2Vec
sentences = [["cat", "say", "meow"] ,["dog", "say", "woof"]] 
```

```python
# Calcule du word embedding 
model = Word2Vec(sentences, min_count=1)
```

```python
model.wv['cat']
```

```python
# Recherche de similarité entre les mots
model.wv.most_similar('cat')
```
