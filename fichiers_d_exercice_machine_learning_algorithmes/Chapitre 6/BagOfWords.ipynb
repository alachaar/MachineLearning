```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# 1 -Creation d'un sac de mots vides
vectorizer = CountVectorizer()
```

```python
# 2 -Creation d'un corpus (3 documents)
corpus = ['un jour ici et la','demain mais pas ici','la la la lere']
```

```python
# Calcul d'un sac
vectorizer.fit(corpus)
bow = vectorizer.transform(corpus)
```

```python
# Affichage de tous les mots du sac
vectorizer.get_feature_names_out()
```

```python
# Affichage brut du denombrement des mots
print(bow.toarray())
```

```python
# Affichage dans un tableau
import pandas as pd
pd.DataFrame(bow.toarray(), columns=vectorizer.get_feature_names_out())
```
