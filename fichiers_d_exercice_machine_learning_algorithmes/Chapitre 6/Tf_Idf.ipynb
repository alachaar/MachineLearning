```python
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
corpus = ['un jour ici et la','demain mais pas la la la ici','la la la la la la la la la lere']
```

```python
# contage
vectorizer = CountVectorizer()
counts = vectorizer.fit_transform(corpus)
counts
```

```python
# Calcul TF-IDF a partir du contage
transformer = TfidfTransformer()
transformer.fit_transform(counts)
transformer.idf_
```

```python
dic = dict(zip(vectorizer.get_feature_names_out(), transformer.idf_))
pd.DataFrame.from_dict(dic,orient='index')
```
